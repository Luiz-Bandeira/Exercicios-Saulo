## Como resolver um problema de programação?

1.  **Entenda o problema:** Leia com atenção o que está sendo pedido.
2.  **Identifique as Entradas e Saídas:** O que eu recebo do usuário? O que eu devo mostrar na tela?
3.  **Esboce o Passo a Passo (Algoritmo):** "Primeiro, peço o valor X. Depois verifico se X é maior que 10. Se sim, faço isso."
4.  **Codifique:** Transforme seu passo a passo em código Python.
5.  **Teste:** Execute com os valores de exemplo e veja se funciona.

## Exemplo Resolvido 1: Ano Bissexto

**O problema:** Crie um programa que peça um ano (ex: 2024) e informe se ele é Bissexto ou não.

*Regra do Ano Bissexto:*
Ele precisa ser múltiplo de 4 (`ano % 4 == 0`).
Mas se for fim de século (múltiplo de 100), ele **não** é bissexto, a menos que ele também seja múltiplo de 400.

```python
# Solução do Professor
ano = int(input("Digite o ano: "))

if (ano % 4 == 0 and ano % 100 != 0) or (ano % 400 == 0):
    print(f"O ano {ano} É bissexto!")
else:
    print(f"O ano {ano} NÃO é bissexto.")
```

## Exemplo Resolvido 2: Calculadora Simples

**O problema:** Peça dois números e um operador matemático (`+`, `-`, `*` ou `/`) ao usuário. Imprima o resultado da operação.

```python
num1 = float(input("Digite o primeiro número: "))
op = input("Digite a operação (+, -, *, /): ")
num2 = float(input("Digite o segundo número: "))

if op == "+":
    print(f"Resultado: {num1 + num2}")
elif op == "-":
    print(f"Resultado: {num1 - num2}")
elif op == "*":
    print(f"Resultado: {num1 * num2}")
elif op == "/":
    # Cuidados extras em programação: não se divide por zero!
    if num2 == 0:
        print("Erro: Divisão por zero.")
    else:
        print(f"Resultado: {num1 / num2}")
else:
    print("Operador inválido!")
```

### Desafio 3: Triângulos
Peça os 3 lados de um triângulo. O programa deverá informar se os valores formam um triângulo. Caso formem, diga o tipo do triângulo:
- Equilátero: 3 lados iguais;
- Isósceles: 2 lados iguais;
- Escaleno: 3 lados diferentes.
*Dica: Para formar um triângulo, a soma de quaisquer dois lados precisa ser maior que o terceiro lado.*

```python
lado1 = input("Digite o primeiro lado:")
lado2 = input("Digite o segundo lado:")
lado3 = input("Digite o terceiro lado:")


if lado1 == lado2 == lado3:
  print("Equilátero")
elif lado1 != lado2 != lado3:
    print("Escaleno")
else:
  print("Isósceles")
```

## Exercício 9 (C): Verificação de Faixa e Condição Múltipla
Escreva um programa em C que leia um número inteiro. Se o número estiver no intervalo de 10 a 50 (inclusive) E for divisível por 3, imprima 'Condição A atendida'. Se o número for menor que 10 OU maior que 100, E não for divisível por 5, imprima 'Condição B atendida'. Caso contrário, imprima 'Nenhuma condição atendida'.

```python
numero_inteiro = int(input("Digite um número inteiro: "))

if 10 <= numero_inteiro <= 50:
  if numero_inteiro % 3 == 0:
    print("Condição A atendida")
elif numero_inteiro < 10 or numero_inteiro > 100:
  if numero_inteiro % 5 == 0:
    print("Condição B atendida")
  else:
    print("Nenhuma condição atendida")

```

## Exercício 10 (C): Prioridade em Portaria Complexa
Uma portaria de evento tem regras de entrada complexas. Crie um programa em C que leia a idade do visitante (`int idade`), se ele possui ingresso (`char tem_ingresso` - 'S' ou 'N') e se ele é VIP (`char is_vip` - 'S' ou 'N').
As regras de entrada são:
1. Se a idade for menor que 18, a entrada é negada.
2. Se a idade for 18 ou mais:
   a. Se for VIP, a entrada é liberada, independentemente de ter ingresso.
   b. Se não for VIP, mas tiver ingresso, a entrada é liberada.
   c. Se não for VIP e não tiver ingresso, a entrada é negada.

Imprima 'Entrada Liberada!' ou 'Entrada Negada!' conforme as regras. Valide as entradas para `tem_ingresso` e `is_vip` para que aceitem apenas 'S' ou 'N' (case-insensitive).

```python
idade = int(input("Digitee sua idade:"))
tem_ingresso = input("Você possui ingresso? (S/N)").upper()
is_vip = input("Você é VIP? (S/N) ")

if idade < 18:
  print("Entrada Negada")
elif idade >= 18 and is_vip == "S":
  print("Entrada Liberada")
elif idade >= 18 and is_vip == "N" and tem_ingresso == "S":
  print("Entrada Liberada")
elif idade >= 18 and is_vip == "N" and tem_ingresso == "N":
  print("Entrada Negada")
else:
  print("Erro")
```
