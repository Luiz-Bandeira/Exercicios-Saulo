## 1. Arquivos de Texto Simples `.txt`

O Python possui a função `open()` para abrir arquivos. Precisamos dizer o nome do arquivo e o **Modo** de abertura:
- `'r'` (read): Apenas para Ler (padrão).
- `'w'` (write): Escrever por cima de tudo.
- `'a'` (append): Escrever no final do arquivo.

```python
# Escrevendo em um arquivo (vai ser criado nesta mesma pasta)
arquivo = open("meu_diario.txt", "w")
arquivo.write("Linha 1.\n")
arquivo.write("Linha 2.")
arquivo.write("Linha 3.\n")
arquivo.write("Linha 4.")
arquivo.close()
```

```python
arquivo = open("meu_diario.txt", "w")
arquivo.write("Hoje comecei a aprender sobre aqrquivos em Phyton.\n")
arquivo.close()
```

### O bloco `with`
Esquecer o `arquivo.close()` é comum e pode corromper os dados. A forma profissional de fazer isso no Python é usando o bloco `with`. Ele abre e **garante que vai fechar**.

```python
# Lendo o arquivo usando with (Muito mais seguro e limpo!)
with open("meu_diario.txt", "r") as arquivo:
    conteudo = arquivo.read()
    print("Li do arquivo o seguinte texto:\n" + conteudo)

# Já está fechado aqui fora!
```

## 2. Arquivos JSON

O `.txt` é muito básico. E se quisermos salvar um Dicionário ou uma Lista complexa?
O formto **JSON** (JavaScript Object Notation) é o padrão de mercado para isso na internet. Ele é basicamente texto, mas formatado exatamente como os Dicionários e Listas do Python!

```python
import json

meus_dados = {
    "nome": "Saulo",
    "habilidades": ["Python", "Dados"],
    "nivel": 10
}

# Salvando um Dicionário Python para um arquivo JSON (DUMP)
with open("perfil.json", "w") as f:
    json.dump(meus_dados, f, indent=4)

print("Arquivo JSON criado! Abra a pasta e dê um duplo clique no perfil.json para olhar.")
```

```python
# Lendo um JSON devolta para o Python (LOAD)
with open("perfil.json", "r") as f:
    dados_carregados = json.load(f)

print(f"Bem-vindo de volta, {dados_carregados['nome']}!")
```

## 3. Arquivos CSV (Excel / Planilhas)

CSV (Comma Separated Values) é o formato padrão para exportar tabelas. São como planilhas do Excel, onde cada linha do arquivo é uma linha da tabela, e as colunas são separadas por vírgula `,` ou ponto-e-vírgula `;`.
O Python possui o módulo nativo `csv` para ajudar.

```python
import csv

# Escrevendo dados tabulares
alunos = [
    ["Nome", "Idade", "Nota"],
    ["Maria", 22, 9.5],
    ["José", 25, 7.0]
]

with open("turma.csv", "w", newline='', encoding='utf-8') as f:
    escritor = csv.writer(f)
    escritor.writerows(alunos)

print("Arquivo da turma criado!")
```

```python
'''
Prompt:
Crie um programa em python que me permita entender
o funcionamento de arquivos tipo texto.
O programa deve possuir um menu de opcoes para
incluir, listar, alterar e eliminar logicamente
registros dentro deste arquivo. Suponha que a
estrutura deste arquivo sera formada por: "codigo", "nome", "idade" e "flag". O campo flag sera usado para efetuar eliminacao logica.
Se o valor for True o registro eh valido se False
o registro esta eliminado.
'''

# ============================================================
# SISTEMA DE CADASTRO EM ARQUIVO TEXTO
# ============================================================
#
# Objetivo:
# Demonstrar o funcionamento de arquivos texto em Python.
#
# Cada registro possui:
# codigo;nome;idade;flag
#
# Exemplo:
# 1;Joao;20;True
#
# flag = True  -> registro válido
# flag = False -> eliminado logicamente
#
# ============================================================

ARQUIVO = "cadastro.txt"


# ============================================================
# FUNÇÃO: incluir registro
# ============================================================

def incluir():

    codigo = input("Codigo: ")
    nome = input("Nome: ")
    idade = input("Idade: ")

    registro = f"{codigo};{nome};{idade};True\n"

    with open(ARQUIVO, "a", encoding="utf-8") as arq:
        arq.write(registro)

    print("\nRegistro incluído com sucesso!\n")


# ============================================================
# FUNÇÃO: listar registros válidos
# ============================================================

def listar():

    try:

        with open(ARQUIVO, "r", encoding="utf-8") as arq:

            print("\n===== REGISTROS =====\n")

            encontrou = False

            for linha in arq:

                linha = linha.strip()

                dados = linha.split(";")

                codigo = dados[0]
                nome = dados[1]
                idade = dados[2]
                flag = dados[3]

                # Exibe somente registros válidos
                if flag == "True":

                    encontrou = True

                    print(f"Codigo : {codigo}")
                    print(f"Nome   : {nome}")
                    print(f"Idade  : {idade}")
                    print("---------------------")

            if not encontrou:
                print("Nenhum registro válido encontrado.")

            print()

    except FileNotFoundError:
        print("\nArquivo ainda não existe.\n")


# ============================================================
# FUNÇÃO: alterar registro
# ============================================================

def alterar():

    codigo_busca = input("Digite o codigo a alterar: ")

    novos_registros = []

    encontrado = False

    try:

        with open(ARQUIVO, "r", encoding="utf-8") as arq:

            for linha in arq:

                linha = linha.strip()

                dados = linha.split(";")

                codigo = dados[0]
                nome = dados[1]
                idade = dados[2]
                flag = dados[3]

                # Altera apenas se registro existir e estiver válido
                if codigo == codigo_busca and flag == "True":

                    print("\nRegistro encontrado!")

                    novo_nome = input("Novo nome: ")
                    nova_idade = input("Nova idade: ")

                    nova_linha = f"{codigo};{novo_nome};{nova_idade};True\n"

                    novos_registros.append(nova_linha)

                    encontrado = True

                else:
                    novos_registros.append(linha + "\n")

        if encontrado:

            with open(ARQUIVO, "w", encoding="utf-8") as arq:

                arq.writelines(novos_registros)

            print("\nRegistro alterado com sucesso!\n")

        else:
            print("\nRegistro não encontrado.\n")

    except FileNotFoundError:
        print("\nArquivo não existe.\n")


# ============================================================
# FUNÇÃO: exclusão lógica
# ============================================================

def excluir():

    codigo_busca = input("Digite o codigo a excluir: ")

    novos_registros = []

    encontrado = False

    try:

        with open(ARQUIVO, "r", encoding="utf-8") as arq:

            for linha in arq:

                linha = linha.strip()

                dados = linha.split(";")

                codigo = dados[0]
                nome = dados[1]
                idade = dados[2]
                flag = dados[3]

                # Exclusão lógica
                if codigo == codigo_busca and flag == "True":

                    nova_linha = f"{codigo};{nome};{idade};False\n"

                    novos_registros.append(nova_linha)

                    encontrado = True

                else:
                    novos_registros.append(linha + "\n")

        if encontrado:

            with open(ARQUIVO, "w", encoding="utf-8") as arq:

                arq.writelines(novos_registros)

            print("\nRegistro eliminado logicamente.\n")

        else:
            print("\nRegistro não encontrado.\n")

    except FileNotFoundError:
        print("\nArquivo não existe.\n")


# ============================================================
# MENU PRINCIPAL
# ============================================================

while True:

    print("===================================")
    print(" SISTEMA DE ARQUIVO TEXTO")
    print("===================================")
    print("1 - Incluir")
    print("2 - Listar")
    print("3 - Alterar")
    print("4 - Excluir logicamente")
    print("5 - Sair")
    print("===================================")

    opcao = input("Escolha uma opção: ")

    if opcao == "1":
        incluir()

    elif opcao == "2":
        listar()

    elif opcao == "3":
        alterar()

    elif opcao == "4":
        excluir()

    elif opcao == "5":
        print("\nEncerrando sistema...")
        break

    else:
        print("\nOpção inválida!\n")
```
